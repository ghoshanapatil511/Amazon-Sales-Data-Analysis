# **Step 1: Data Loading and Understanding**

```python
import pandas as pd
import numpy as np


df = pd.read_csv('/content/amazon.csv')

df.head()
```

# **Missing Data Treatment**

```python

df['rating_count'] = pd.to_numeric(
    df['rating_count'].astype(str).str.replace(',', ''), errors='coerce'
)


df['rating_count'] = df['rating_count'].fillna(df['rating_count'].median())
```

```python
df.isna().sum()
```

# **Step 3:-Data Formatting and Type Conversion**



```python
df['discounted_price'] = pd.to_numeric(
    df['discounted_price'].astype(str).str.replace(',', '').str.replace('₹', '').str.strip(),
    errors='coerce'
)

df['actual_price'] = pd.to_numeric(
    df['actual_price'].astype(str).str.replace(',', '').str.replace('₹', '').str.strip(),
    errors='coerce'
)

df['discount_percentage'] = pd.to_numeric(
    df['discount_percentage'].astype(str).str.replace('%', '').str.strip(),
    errors='coerce'
)

df['rating'] = pd.to_numeric(df['rating'], errors='coerce')
df['rating_count'] = pd.to_numeric(
    df['rating_count'].astype(str).str.replace(',', '').str.strip(),
    errors='coerce'
)
```

# **Step 5:-Exploratory Data Analysis (EDA)**

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

### **Q1. What is the distribution of product ratings?**

```python

plt.figure(figsize=(10, 5))
sns.histplot(df['rating'], bins=20, kde=True, color='blue')
plt.title('Distribution of Product Ratings')
plt.xlabel('Rating')
plt.ylabel('Frequency')
plt.show()
```

**Analysis:**

Most ratings are skewed towards the higher end (between 3.5 and 4.5).

* Most ratings are skewed towards the higher end (between 3.5 and 4.5).

* Very few products have ratings below 3.0.

**Conclusion:**
*   Majority of the products on Amazon have positive customer ratings.




### **Q2. How does discount percentage affect product ratings?**

```python
plt.figure(figsize=(10, 5))
sns.scatterplot(
    x='discount_percentage', y='rating', data=df, color='green', alpha=0.6
)
plt.title('Impact of Discount Percentage on Rating')
plt.xlabel('Discount Percentage (%)')
plt.ylabel('Rating')
plt.show()
```

**Analysis:**
*   Ratings are evenly distributed across all discount percentages with no clear trend.


**Conclusion:**
Higher discounts do not guarantee better product ratings.

*  Higher discounts do not guarantee better product ratings.





### **Q3. What are the top 10 categories with the maximum number of products?**

```python
top_categories = df['category'].value_counts().head(10)

plt.figure(figsize=(10, 6))
top_categories.plot(kind='barh', color='teal')
plt.title('Top 10 Categories with Maximum Number of Products')
plt.xlabel('Number of Products')
plt.ylabel('Category')
plt.gca().invert_yaxis()  # सबसे बड़ी कैटेगरी को ऊपर दिखाने के लिए
plt.show()
```

**Analysis:**


*   Electronics and Home & Kitchen have significantly higher product counts than other categories.




**Conclusion:**
*   Electronics is the most dominant product category in terms of overall inventory.




### **Q4. What is the proportion of products across top categories?**

```python
# Category Distribution
counts = df['category'].value_counts().head(5)

plt.figure(figsize=(10, 5))
plt.pie(counts.values,
        labels=counts.index,
        autopct='%1.2f%%',
        startangle=90,
        shadow=True,
        explode=[0.1, 0, 0, 0, 0],
        colors=['red', 'blue', 'green', 'yellow', 'orange']
)
plt.title('Top 5 Product Categories Share')
plt.show()
```

**Analysis:**

Top 5 categories represent a major share of the total dataset inventory.  
**Conclusion:**

A small group of core categories drives most of the marketplace offerings.

### **Q5. What is the relationship between actual price and discounted price?**

```python
plt.figure(figsize=(10, 5))
sns.violinplot(x=df['discount_percentage'], color='skyblue')
plt.title('Distribution of Discount Percentage (Violin Plot)')
plt.xlabel('Discount Percentage (%)')
plt.show()
```

**Analysis:**

Discount percentages show dense concentration around specific median values.


* Discount percentages show dense concentration around specific median values.


**Conclusion:**
*  Most products follow standard, structured discounting tiers.
